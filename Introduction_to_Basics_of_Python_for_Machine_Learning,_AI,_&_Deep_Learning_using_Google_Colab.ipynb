```python
#2. Variables and Data Types
# Storing different types of data in variables
student_name = "Arjun"      # String (Text)
student_age = 22            # Integer (Whole number)
course_fee = 4999.90        # Float (Decimal number)
is_enrolled = True          # Boolean (True/False)
```

```python
#3. Checking Data Types
# Checking the type of the variables defined earlier
print(type(student_name))  # Output: <class 'str'>
print(type(student_age))   # Output: <class 'int'>
print(type(course_fee))    # Output: <class 'float'>
print(type(is_enrolled))   # Output: <class 'bool'>
```

```python
#4. Taking User Input
# Taking input from the user
input_name = input("Please, Enter your name: ")
print("Hello, " + input_name)
```

```python
#5. Type Conversion
# Taking age as input (as a string) and converting it to an integer
age_input = input("Enter your age: ")
converted_age = int(age_input)

# Performing arithmetic with the converted type
print("Next year, your age will be", converted_age + 1)
```

```python
#6. Basic Arithmetic Operations
num1 = 20
num2 = 7

# Basic arithmetic
addition_result = num1 + num2
subtraction_result = num1 - num2
multiplication_result = num1 * num2
division_result = num1 / num2

print(addition_result, subtraction_result, multiplication_result, division_result)
```

```python
#7.Conditional Statements
marks = 85

# Applying conditions based on the value of 'marks'
if marks >= 90:
    print("Excellent performance")
elif marks >= 75:
    print("Good job")
else:
    print("Keep improving")
```

```python
#8.For Loop
# 'range(1, 6)' generates numbers 1, 2, 3, 4, 5
for number in range(1, 6):
    print("Current number:", number)
```

```python
counter_value = 1

# Runs until the condition is no longer true
while counter_value <= 5:
    print("Iteration:", counter_value)
    counter_value += 1  # Increment to prevent an infinite loop
```

```python
#9. Functions
# Defining a function to calculate square
def calculate_square(input_number):
    result = input_number ** 2
    return result

# Calling the function
output_value = calculate_square(8)
print("The square is", output_value)
```
